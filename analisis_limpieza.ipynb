```python
import numpy as np

def limpiar_moneda(serie):
    # 1. Convertimos a texto por las dudas, reemplazamos valores nulos por texto vacío temporalmente
    serie_str = serie.astype(str).str.replace("nan", "", case=False)
    
    # 2. Sacamos las letras (como 'ARS') y los símbolos '$'
    serie_limpia = (
        serie_str
        .str.replace("ARS", "", case=False, regex=False)
        .str.replace("$", "", regex=False)
    )
    
    # 3. Manejamos los puntos y comas de manera inteligente:
    # Si tiene un punto como miles y una coma como decimal (ej: 1.500,50)
    # Primero sacamos los puntos de miles y después cambiamos la coma decimal por punto.
    serie_limpia = (
        serie_limpia
        .str.replace(".", "", regex=False)    # Saca el punto de miles
        .str.replace(",", ".", regex=False)  # Cambia la coma por punto para que Python/SQL lo entienda como decimal
        .str.strip()                         # Saca espacios vacíos sobrantes
    )
    
    # 4. Convertimos a numérico: lo que no se pueda convertir (como celdas vacías), lo transforma en NaN (número nulo)
    return pd.to_numeric(serie_limpia, errors="coerce")

# Aplicamos la función a tus dos columnas problemáticas
df_datosSucios["precio_unitario"] = limpiar_moneda(df_datosSucios["precio_unitario"])
df_datosSucios["monto_total"] = limpiar_moneda(df_datosSucios["monto_total"])

df_datosSucios.head(10)
```

```python
# 1. Limpiamos espacios, pasamos a formato Título (primera letra en mayúscula)
df_datosSucios["cliente"] = (
    df_datosSucios["cliente"]
    .astype(str)
    .str.strip()
    .str.title()
)

# 2. Mostramos cómo quedó la columna cliente actualizada
print(df_datosSucios[["id_venta", "cliente"]])
```

```python
import unicodedata

def limpiar_nombre_cliente(texto):
    if pd.isna(texto):
        return texto
    
    # 1. Normaliza el texto para separar las letras de sus tildes
    nfkd_form = unicodedata.normalize('NFKD', str(texto))
    
    # 2. Filtra y elimina los caracteres de acentuación (combining characters)
    sin_tildes = "".join([c for c in nfkd_form if not unicodedata.combining(c)])
    
    # 3. Devuelve el texto limpio, sin espacios extra y en formato título (Ej: "Maria Gomez")
    return sin_tildes.strip().title()

# Aplicamos la función a la columna cliente
df_datosSucios["cliente"] = df_datosSucios["cliente"].apply(limpiar_nombre_cliente)

# Verificamos cómo quedó la columna
print(df_datosSucios[["id_venta", "cliente"]])
```

```python
import unicodedata

def limpiar_provincia(texto):
    if pd.isna(texto):
        return texto  # Si es NaN, lo deja tranquilo para tratarlo después
    
    # 1. Quitar tildes
    nfkd_form = unicodedata.normalize('NFKD', str(texto))
    sin_tildes = "".join([c for c in nfkd_form if not unicodedata.combining(c)])
    
    # 2. Limpiar espacios y aplicar formato título
    return sin_tildes.strip().title()

# Aplicamos la función base para quitar tildes y normalizar formato
df_datosSucios["provincia"] = df_datosSucios["provincia"].apply(limpiar_provincia)

# 3. Diccionario de mapeo para unificar abreviaturas y variantes de provincias
mapeo_provincias = {
    "Mza.": "Mendoza",
    "Mza": "Mendoza",
    "Mdz": "Mendoza",
    "Bs. As.": "Buenos Aires",
    "Bs As": "Buenos Aires",
    "Bsas": "Buenos Aires"
}

df_datosSucios["provincia"] = df_datosSucios["provincia"].replace(mapeo_provincias)

# Verificamos los valores únicos de provincias para ver cómo quedó
print(df_datosSucios["provincia"].value_counts(dropna=False))
```

```python
# Automatizar: Si el cliente es Maria Gomez y la provincia es NaN, ponerle Santa Fe en cualquier parte de la tabla
df_datosSucios.loc[
    (df_datosSucios["cliente"] == "Maria Gomez") & (df_datosSucios["provincia"].isna()), 
    "provincia"
] = "Santa Fe"


df_datosSucios.head(20)
```

```python
# Automatizar: Si el cliente es Corralon San Jose y la provincia es NaN, ponerle Buenos Aires
df_datosSucios.loc[
    (df_datosSucios["cliente"] == "Corralon San Jose") & (df_datosSucios["provincia"].isna()), 
    "provincia"
] = "Buenos Aires"


df_datosSucios.head(20)
```

```python
# Eliminar las filas que no tienen fecha de venta (los NaT)
df_datosSucios = df_datosSucios.dropna(subset=["fecha_venta"])

# Verificamos cuántas filas nos quedaron limpias
print(f"Total de filas limpias y listas: {len(df_datosSucios)}")


df_datosSucios.head(15)
```

```python
import pandas as pd
from sqlalchemy import create_engine

# 1. Nos conectamos a nuestra base de datos SQLite existente
engine = create_engine('sqlite:///ventas_ferreteria.db')

query_evolucion_mes = """
    SELECT 
        strftime('%Y-%m', fecha_venta) AS mes_venta,
        COUNT(id_venta) AS cantidad_ventas,
        SUM(monto_total) AS facturacion_mensual,
        ROUND(AVG(monto_total), 2) AS ticket_promedio_mensual
    FROM tb_ventas
    GROUP BY mes_venta
    ORDER BY mes_venta ASC;
"""

df_evolucion = pd.read_sql(query_evolucion_mes, con=engine)
df_evolucion
```
